## 1. Create a 1D NumPy Array

A 1D NumPy array named `temperatures_w1` is created to store the daily average temperatures for Week 1.

```python
import numpy as np
temperatures_w1=np.array([22.5,23.3,20.8,23.4,26.1,24.8,21.9])
print(temperatures_w1)
```

## 2. Inspection and Properties

The shape, data type, and number of elements in the Week 1 temperature array are inspected using NumPy properties.

```python
print("Shape:",temperatures_w1.shape)
print("Data type:",temperatures_w1.dtype)
print("Number of elements:",temperatures_w1.size)
```

## 3. Convert Celsius to Fahrenheit

The temperatures are converted from Celsius to Fahrenheit using the formula:
Fahrenheit = (Celsius × 9/5) + 32

```python
temperatures_f=(temperatures_w1*9/5)+32
print(temperatures_f)
```

### Maximum, Minimum and Mean Temperature

```python
print("Maximum temperature:",temperatures_w1.max())
print("Minimum temperature:",temperatures_w1.min())
print("Mean temperature:",temperatures_w1.mean())
```

## 4. Array Slicing and Indexing

Array slicing is used to extract temperatures for specific days of the week.

```python
first_three_days=temperatures_w1[:3]
print("First 3 days:",first_three_days)
week_end=temperatures_w1[-2:]
print("Week end:",week_end)
middle_three_days=temperatures_w1[2:5]
print("Middle 3 days:",middle_three_days)
```

## 5. Create a 2D NumPy Array

A 2D NumPy array named `temperatures` is created.

Each row represents the temperatures recorded during a different week.

## 6 Inspect and Slice the 2D Array

Inspect the shape, data type, and total number of elements in the 2D array.

Extract:

- Weekend temperatures for both weeks

# Pandas Series

A Pandas Series is a one-dimensional labeled data structure.

## 1. Creating a Pandas Series

A Pandas Series named `marks` is created with custom index labels representing student ranks.

```python
import pandas as pd
marks=pd.Series([95,92,89,85,80],index=['Rank1','Rank2','Rank3','Rank4','Rank5'])
print(marks)
```

## 2.2 Indexing and Slicing

Perform the following operations:

- Access the mark of the 1st rank student using integer position
- Retrieve the top 3 ranks using `loc`
- Retrieve the 3rd-ranked student's mark using `iloc`
- Filter students whose marks are greater than 90

```python
# Mark of first rank student
print("Marks of first rank student",marks.iloc[0])

# Top 3 ranks
print("\nTop 3 ranks\n",marks.loc[['Rank1','Rank2','Rank3']])

# 3rd rank student
print("\nMark of third rank student:",marks.iloc[2])

# Marks greater than 90
print("\nMarks greater than 90\n",marks[marks>90])
```

## 2.3 Manipulating the Series

Perform the following operations:

- Modify the 1st rank student's mark to 100
- Remove the entry corresponding to Rank5
- Calculate CGPA by dividing each mark by 10

```python
# Modify 1st rank mark
marks.loc['Rank1']=100
print(marks)

# Remove last rank
marks=marks.drop('Rank5')
print("\nAfter removing the last rank:\n",marks)

# CGPA
cgpa=marks/10
print("\nCGPA:\n",cgpa)
```

# 3. Pandas DataFrame

In this section, a transaction dataset is created and explored using Pandas DataFrame operations.

## 3.1 Create the Transactions DataFrame

Create a DataFrame named `transactions` containing TransactionID, ProductCategory, Region, and Amount.

```python
transactions = pd.DataFrame({
    'TransactionID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                       'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region':['North', 'South', 'North', 'East', 'West',
              'North', 'East', 'West', 'South', 'North'],
    'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

## 3.2 Data Exploration

Explore the transactions DataFrame using:

- `head()`
- `tail()`
- `shape`
- `columns`
- `dtypes`

```python
print("First 5 rows:\n",transactions.head())
print("\nLast 5 rows:\n",transactions.tail())
print("\nShape:\n",transactions.shape)
print("\nColumns:\n",transactions.columns)
print("\nDatatypess:\n",transactions.dtypes)
```

## 3.3 Select Specific Columns

Display only the `ProductCategory` and `Amount` columns.

## 3.4 Retrieve the Last Three Columns

Use `iloc` or `loc` to retrieve the last three columns of the DataFrame.

```python
print(transactions[['ProductCategory','Amount']])

# Last 3 columns
print("\nLast 3 columns:\n")
print(transactions.iloc[:,-3:])
```

## 3.5 Filter Transactions

Filter the transactions where:

- Region is `North`
- Amount is greater than 200

```python
filtered_transactions=transactions[(transactions['Region']=='North') & (transactions['Amount']>200)]
print(filtered_transactions)
```

## 3.6 Analyze Product Categories

Find the number of transactions for each `ProductCategory` using `value_counts()`.

```python
print(transactions['ProductCategory'].value_counts())
```

## 3.7 Find Unique Regions

Find the unique values present in the `Region` column.

```python
print(transactions['Region'].unique())
```

## 3.8 Group Transactions by Region

Group the transactions by `Region` and calculate the mean `Amount` for each region.

```python
mean_regions=transactions.groupby('Region')['Amount'].mean()
print(mean_regions)
```

# 4. DataFrame Manipulation

In this section, the transactions DataFrame is modified by updating values, adding columns, and removing rows and columns.

## 4.1 Modify Transaction Amount

Modify the Amount for TransactionID 102 from 150 to 165.

```python
transactions.loc[transactions['TransactionID']==102,'Amount']=165
print(transactions)
```

## 4.2 Add a Discount Column

Add a new column named `Discount` by calculating 10% of the transaction Amount.

```python
transactions['Discount']=transactions['Amount']*0.10
print(transactions)
```

## 4.3 Remove TransactionID 109

Remove the row corresponding to TransactionID 109 from the DataFrame.

```python
transactions=transactions.drop(transactions[transactions['TransactionID']==109].index)
print(transactions)
```

## 4.4 Delete the Discount Column

Delete the `Discount` column from the DataFrame.

```python
transactions=transactions.drop(columns=['Discount'])
print(transactions)
```
